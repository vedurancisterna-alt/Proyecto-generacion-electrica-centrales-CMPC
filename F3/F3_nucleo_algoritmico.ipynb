### Eficiencia y optimización

Se comparan dos implementaciones de la transformación ancho→largo: la vectorizada (pandas.melt, usada en producción) y una iterativa (iterrows), verificando primero su equivalencia funcional y luego midiendo tiempo (timeit) y memoria (tracemalloc) sobre las 729 filas reales del proyecto.

```python
import re
import timeit
import tracemalloc

def transformar_ancho_largo_iterativo(datos, columnas_base, columnas_hora):
    """Misma transformación, pero con un bucle manual en vez de melt."""
    filas = []
    for _, fila in datos.iterrows():
        base = {col: fila[col] for col in columnas_base}
        for columna_hora in columnas_hora:
            numero_hora = int(re.search(r"(\d+)", columna_hora).group(1))
            nueva_fila = dict(base)
            nueva_fila["Hora"] = numero_hora
            nueva_fila["Generacion_MWh"] = pd.to_numeric(fila[columna_hora], errors="raise")
            filas.append(nueva_fila)
    resultado = pd.DataFrame(filas)
    return resultado[columnas_base + ["Hora", "Generacion_MWh"]]

# Datos reales filtrados a las 3 centrales (729 filas anchas)
datos_729 = FiltradorCentrales(centrales).ajustar(datos_crudos).transformar(datos_crudos)

# 1. Verificar equivalencia
from src.transformacion import transformar_ancho_largo
a = transformar_ancho_largo(datos_729, columnas_base, columnas_hora).sort_values(columnas_base+["Hora"]).reset_index(drop=True)
b = transformar_ancho_largo_iterativo(datos_729, columnas_base, columnas_hora).sort_values(columnas_base+["Hora"]).reset_index(drop=True)
pd.testing.assert_frame_equal(a, b, check_dtype=False)
print("Ambas formas producen el mismo resultado.\n")

# 2. Medir tiempo
rep = 30
t_v = timeit.timeit(lambda: transformar_ancho_largo(datos_729, columnas_base, columnas_hora), number=rep) / rep
t_i = timeit.timeit(lambda: transformar_ancho_largo_iterativo(datos_729, columnas_base, columnas_hora), number=rep) / rep
print(f"melt (vectorizado): {t_v:.6f} s")
print(f"iterrows (bucle)  : {t_i:.6f} s")
print(f"El vectorizado es {t_i/t_v:,.1f} veces más rápido\n")

# 3. Medir memoria
tracemalloc.start(); transformar_ancho_largo(datos_729, columnas_base, columnas_hora); _, p_v = tracemalloc.get_traced_memory(); tracemalloc.stop()
tracemalloc.start(); transformar_ancho_largo_iterativo(datos_729, columnas_base, columnas_hora); _, p_i = tracemalloc.get_traced_memory(); tracemalloc.stop()
print(f"Memoria melt    : {p_v/1024/1024:.4f} MB")
print(f"Memoria iterrows: {p_i/1024/1024:.4f} MB")
```

### Comparación a escala completa: implementación iterativa y vectorizada

Como complemento a la comparación iterativa–recursiva, se evalúan dos implementaciones capaces de procesar la totalidad del dataset de F2, compuesto por 17.496 observaciones.

El objetivo es comparar el comportamiento de una solución iterativa en Python frente a una solución vectorizada basada en pandas para identificar registros con generación igual a 0 MWh. Ambas implementaciones resuelven exactamente la misma tarea y su equivalencia funcional se verifica antes de medir el rendimiento.

Esta comparación permite analizar la eficiencia sobre el volumen real del proyecto sin la restricción de profundidad de pila presente en la implementación recursiva.

```python
import timeit
import tracemalloc
import pandas as pd


def detectar_ceros_iterativo(df):
    """Identifica registros con generación igual a 0 MWh mediante iteración."""
    resultado = []

    for valor in df["Generacion_MWh"]:
        resultado.append(valor == 0)

    return resultado


def detectar_ceros_vectorizado(df):
    """Identifica registros con generación igual a 0 MWh mediante pandas."""
    return df["Generacion_MWh"].eq(0).tolist()


# Dataset completo de F2
datos_benchmark = datos_finales.copy()

print("Registros evaluados:", len(datos_benchmark))


# 1. Verificación de equivalencia funcional
resultado_iterativo = detectar_ceros_iterativo(datos_benchmark)
resultado_vectorizado = detectar_ceros_vectorizado(datos_benchmark)

assert resultado_iterativo == resultado_vectorizado

print("Equivalencia funcional: OK")
print("Registros con 0 MWh:", sum(resultado_iterativo))


# 2. Medición de tiempo
repeticiones = 100

tiempo_iterativo = timeit.timeit(
    lambda: detectar_ceros_iterativo(datos_benchmark),
    number=repeticiones
) / repeticiones

tiempo_vectorizado = timeit.timeit(
    lambda: detectar_ceros_vectorizado(datos_benchmark),
    number=repeticiones
) / repeticiones


# 3. Medición de memoria
def medir_memoria(funcion, datos):
    tracemalloc.start()

    funcion(datos)

    _, memoria_maxima = tracemalloc.get_traced_memory()
    tracemalloc.stop()

    return memoria_maxima


memoria_iterativa = medir_memoria(
    detectar_ceros_iterativo,
    datos_benchmark
)

memoria_vectorizada = medir_memoria(
    detectar_ceros_vectorizado,
    datos_benchmark
)


# 4. Resumen
comparacion_completa = pd.DataFrame({
    "Implementacion": ["Iterativa", "Vectorizada"],
    "Registros": [len(datos_benchmark)] * 2,
    "Tiempo_promedio_s": [
        tiempo_iterativo,
        tiempo_vectorizado
    ],
    "Memoria_maxima_bytes": [
        memoria_iterativa,
        memoria_vectorizada
    ]
})

comparacion_completa
```

### Interpretación de la comparación a escala completa

Sobre las **17.496 observaciones** del dataset procesado, ambas implementaciones produjeron resultados funcionalmente equivalentes e identificaron **4.438 registros con generación igual a 0 MWh**.

La implementación iterativa registró un tiempo promedio de ejecución de aproximadamente **0,003128 segundos**, mientras que la implementación vectorizada alcanzó aproximadamente **0,000243 segundos**. En este entorno de ejecución, la solución vectorizada fue cerca de **12,9 veces más rápida** que la alternativa iterativa.

En cuanto al uso de memoria medido mediante `tracemalloc`, la implementación iterativa alcanzó un máximo de **153.928 bytes** y la vectorizada **158.456 bytes**. Esto representa un aumento aproximado de **2,9 %** en la memoria máxima observada para la solución vectorizada.

Los resultados evidencian un compromiso entre tiempo y espacio: para el volumen completo del proyecto, la implementación vectorizada presenta una ventaja considerable en velocidad, mientras que la implementación iterativa registra un consumo de memoria ligeramente menor. Considerando que la diferencia de memoria observada es reducida frente a la mejora temporal obtenida, la vectorización resulta adecuada para esta operación sobre el dataset completo.

Los valores absolutos de tiempo y memoria dependen del entorno de ejecución. Por ello, estas mediciones se interpretan como una comparación reproducible entre ambas implementaciones bajo las mismas condiciones y no como valores universales de rendimiento.


## Arquitectura orientada a objetos y patrón Strategy

La Fase 3 reorganiza y amplía componentes desarrollados durante F2 mediante principios de Programación Orientada a Objetos (POO). El módulo `nucleo_poo.py` define la clase base `Transformador` y las clases especializadas `FiltradorCentrales`, `TransformadorAnchoLargo` y `ConstructorVariablesDerivadas`.

### Principios POO implementados

**Herencia.** `FiltradorCentrales`, `TransformadorAnchoLargo` y `ConstructorVariablesDerivadas` heredan de `Transformador`. La clase base establece el contrato común mediante los métodos `ajustar()` y `transformar()`, mientras que cada subclase implementa únicamente el comportamiento que distingue su etapa del procesamiento.

**Polimorfismo.** `Pipeline` recorre los distintos transformadores utilizando la misma interfaz `ajustar()` / `transformar()`, sin necesitar consultar el tipo concreto de cada objeto. Esto permite incorporar nuevas transformaciones compatibles sin modificar la lógica general del pipeline.

**Encapsulamiento.** Cada clase concentra el estado y el comportamiento correspondiente a su responsabilidad. Los parámetros y el estado de ajuste se administran dentro de los objetos, evitando que otros componentes deban manipular directamente su lógica interna.

Esta organización favorece alta cohesión, bajo acoplamiento, reutilización y extensibilidad.

### Patrón Strategy

El módulo `agregacion_temporal.py` implementa el patrón **Strategy** mediante `EstrategiaAgregacion` y sus estrategias horaria, diaria, mensual y por día de semana. `CaracterizadorTemporal` utiliza cualquiera de estas estrategias a través de un contrato común. El patrón es pertinente al problema porque el criterio de agregación temporal cambia mientras el contexto de caracterización permanece estable.

## Síntesis de arquitectura y decisiones de diseño

La arquitectura de F3 mantiene una separación explícita de responsabilidades:

- `carga.py`: lectura controlada de la fuente de datos.
- `transformacion.py`: transformación ancho → largo reutilizada desde F2.
- `validacion.py`: reglas independientes de calidad e integridad.
- `nucleo_poo.py`: pipeline orientado a objetos mediante una clase base `Transformador` y transformadores especializados.
- `agregacion_temporal.py`: estrategias intercambiables para caracterización horaria, diaria, mensual y por día de semana.
- `secuencias.py`: algoritmos iterativo y recursivo, pruebas controladas y mediciones de rendimiento.
- `exportacion.py`: apoyo para la exportación de resultados.

El diseño busca **alta cohesión**, al mantener una responsabilidad principal por módulo/clase, y **bajo acoplamiento**, al depender de contratos comunes en lugar de condicionales sobre tipos concretos. La herencia permite especializar transformadores y estrategias; el polimorfismo permite tratarlos mediante interfaces comunes; y el estado interno del pipeline controla su secuencia de ajuste y transformación.

El patrón **Strategy** se utiliza porque el criterio de agregación temporal cambia mientras el contexto de caracterización permanece estable. Esto permite incorporar nuevas agregaciones sin modificar las estrategias existentes ni `CaracterizadorTemporal`.


## Decisiones de eficiencia y complejidad

Se realizaron **tres comparaciones complementarias**:

1. La transformación vectorizada mediante `pandas.melt` frente a una implementación manual basada en `iterrows`. Ambas producen el mismo resultado; la versión vectorizada presenta mejor desempeño en las mediciones realizadas y se conserva como implementación de producción.
2. Las implementaciones iterativa y recursiva para detectar secuencias consecutivas de generación igual a 0 MWh. Ambas presentan complejidad temporal **O(n)**, pero la alternativa recursiva agrega un costo espacial **O(n)** por la pila de llamadas y está limitada por la profundidad máxima de recursión de Python. Por ello, se utiliza sobre tamaños controlados de hasta 750 elementos.
3. Una comparación iterativa–vectorizada sobre las **17.496 observaciones** del dataset completo. En la ejecución registrada, la alternativa vectorizada fue aproximadamente **12,9 veces más rápida**, con un aumento de memoria máxima cercano al **2,9 %**.

Las mediciones con `timeit` y `tracemalloc` complementan el análisis teórico. Los valores absolutos pueden variar entre equipos, por lo que la interpretación se centra en comparar implementaciones bajo las mismas condiciones.


## Conclusiones de Fase 3

1. La reorganización del pipeline de F2 mediante clases preservó la semántica del dataset procesado: la equivalencia se verificó contra el archivo oficial de F2.
2. La arquitectura POO implementa explícitamente **herencia, polimorfismo y encapsulamiento**, permitiendo extender el flujo sin duplicar la lógica ya validada.
3. El patrón **Strategy** permite ejecutar agregaciones horarias, diarias, mensuales y por día de semana bajo un contrato común.
4. Las pruebas normales, límite y de excepción verifican el comportamiento del algoritmo de secuencias antes de aplicarlo al dataset completo.
5. La comparación iterativa–recursiva confirma equivalencia funcional y muestra por qué la solución iterativa es más adecuada para series extensas: evita el costo de pila y la restricción de profundidad de recursión.
6. La comparación a escala completa sobre **17.496 registros** muestra que la solución vectorizada fue aproximadamente **12,9 veces más rápida** que la iterativa en el entorno evaluado, a cambio de un aumento de memoria máxima cercano al **2,9 %**.
7. El análisis identificó **4.438 observaciones con generación igual a 0 MWh** y diferencias descriptivas entre las tres centrales. Estos patrones no permiten atribuir causas operacionales sin información adicional.
8. F3 consolida la continuidad del proyecto: F1 definió y planificó, F2 preparó y validó los datos, y F3 estructuró, comparó y midió el núcleo algorítmico para análisis posteriores.
